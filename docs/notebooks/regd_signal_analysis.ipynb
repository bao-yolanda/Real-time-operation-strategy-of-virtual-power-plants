# RegD统计信息与实际信号对比分析

本 notebook 用于对比历史统计调频信息与研究日的实际 RegD 调频信号。

输出图片：
- `regd_actual_signal_full_day.pdf`
- `regd_distribution_heatmaps.pdf`
- `regd_hourly_mileage_compare.pdf`
- `regd_typical_hour_distribution_compare.pdf`


```python
from pathlib import Path
import sys
from IPython.display import display, Markdown

candidate_roots = [
    Path.cwd(),
    *Path.cwd().parents,
    Path(r"e:\\研究活\\Real-time-operation-strategy-of-virtual-power-plants-main"),
]

PROJECT_ROOT = None
for candidate in candidate_roots:
    if (candidate / "myexp").exists():
        PROJECT_ROOT = candidate
        break

if PROJECT_ROOT is None:
    raise RuntimeError("Could not locate the project root that contains 'myexp'.")

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from myexp.regd_plot_utils import (
    build_regd_stats,
    build_all_regd_figures,
    save_all_regd_figures,
    summarize_stats,
)

DAY_REG = 22
NOF_HIS_DAYS = 14
GRANULARITY = 0.1
OUTPUT_DIR = PROJECT_ROOT / "results" / "regd_signal_figures"

print(f"PROJECT_ROOT = {PROJECT_ROOT}")
print(f"OUTPUT_DIR = {OUTPUT_DIR}")
```

```python
stats = build_regd_stats(
    day_reg=DAY_REG,
    nof_his_days=NOF_HIS_DAYS,
    granularity=GRANULARITY,
    base_dir=PROJECT_ROOT,
)

summary_df = summarize_stats(stats)
summary_df
```

```python
figures = build_all_regd_figures(stats)
for title, fig in figures:
    display(Markdown(f"### {title}"))
    display(fig)
```

```python
saved_files = save_all_regd_figures(stats, OUTPUT_DIR)
print("已保存输出文件：")
for path in saved_files:
    print(path)
```

## 论文中的建议用法

- `regd_actual_signal_full_day.pdf`：展示研究日实际调频信号全天变化。
- `regd_distribution_heatmaps.pdf`：对比逐小时历史统计分布与实际日经验分布。
- `regd_hourly_mileage_compare.pdf`：对比逐小时历史平均里程与实际日里程。
- `regd_typical_hour_distribution_compare.pdf`：展示差异最明显的典型小时分布对比。
